## **Punto A**

***Punto 1***

$$T_{13}^{23} = \frac{V_{23}}{V_{13}} = \frac{V_2 - V_3}{V_1 - V_3} = \frac{[Y_{\text{ind}}]_{13}^{23}}{[Y_{\text{ind}}]_{13}^{13}} = \frac{\begin{vmatrix} 0 & -G_3 \\ -G_2 & G_1+G_2+G_3 \end{vmatrix}}{\begin{vmatrix} G_3 & -G_3 \\ -G_3 & G_1+G_2+G_3 \end{vmatrix}}$$

Numerador ($[Y_{\text{ind}}]_{13}^{23}$): Se suprimen las filas $\{1, 3\}$ y columnas $\{2, 3\}$, conservando las filas $\{2, 4\}$ y columnas $\{1, 4\}$:

$$\begin{vmatrix} 0 & -G_3 \\ -G_2 & G_1+G_2+G_3 \end{vmatrix} = 0 - ((-G_3)(-G_2)) = -G_2 G_3$$

Denominador ($[Y_{\text{ind}}]_{13}^{13}$): Se suprimen las filas $\{1, 3\}$ y columnas $\{1, 3\}$, conservando las filas $\{2, 4\}$ y columnas $\{2, 4\}$:

$$\begin{vmatrix} G_3 & -G_3 \\ -G_3 & G_1+G_2+G_3 \end{vmatrix} = G_3(G_1+G_2+G_3) - (-G_3)^2 = G_3(G_1+G_2)$$

$$\frac{V_{23}}{V_{13}} = \frac{\operatorname{sgn}(1-3)\operatorname{sgn}(2-3)(-1)^{1+3+2+3}(-G_2 G_3)}{G_3(G_1+G_2)} = \frac{G_2 G_3}{G_3(G_1+G_2)} = \frac{G_2}{G_1+G_2}= \frac{1}{s^2 LC + 1}$$


***Punto 2***

$$T_{31}^{21} = \frac{V_{21}}{V_{31}} = \frac{V_2 - V_1}{V_3 - V_1} = \frac{[Y_{\text{ind}}]_{31}^{21}}{[Y_{\text{ind}}]_{31}^{31}} = \frac{\begin{vmatrix} 0 & -G_3 \\ -G_1 & G_1+G_2+G_3 \end{vmatrix}}{\begin{vmatrix} G_3 & -G_3 \\ -G_3 & G_1+G_2+G_3 \end{vmatrix}}$$

Numerador ($[Y_{\text{ind}}]_{31}^{21}$): Se suprimen las filas $\{3, 1\}$ (quedan $\{2, 4\}$) y las columnas $\{2, 1\}$ (quedan $\{3, 4\}$):
  $$\begin{vmatrix} 0 & -G_3 \\ -G_1 & G_1+G_2+G_3 \end{vmatrix} = 0 - ((-G_3)(-G_1)) = -G_1 G_3$$

Denominador ($[Y_{\text{ind}}]_{31}^{31}$): Se suprimen las filas $\{3, 1\}$ y las columnas $\{3, 1\}$ (quedan filas y columnas $\{2, 4\}$):
  $$\begin{vmatrix} G_3 & -G_3 \\ -G_3 & G_1+G_2+G_3 \end{vmatrix} = G_3(G_1+G_2+G_3) - (-G_3)^2 = G_3(G_1+G_2)$$

$$\frac{V_{21}}{V_{31}} = \frac{G_1 G_3}{G_3(G_1+G_2)} = \frac{G_1}{G_1+G_2}= \frac{s^2 LC}{s^2 LC + 1}$$


***Punto 3***

$$T_{32}^{12} = \frac{V_{12}}{V_{32}} = \frac{V_1 - V_2}{V_3 - V_2} = \frac{[Y_{\text{ind}}]_{32}^{12}}{[Y_{\text{ind}}]_{32}^{32}} = \frac{\begin{vmatrix} 0 & -G_2 \\ -G_1 & G_1+G_2+G_3 \end{vmatrix}}{\begin{vmatrix} G_2 & -G_2 \\ -G_2 & G_1+G_2+G_3 \end{vmatrix}}$$

Numerador ($[Y_{\text{ind}}]_{32}^{12}$): Se suprimen las filas $\{3, 2\}$ (quedan $\{1, 4\}$) y las columnas $\{1, 2\}$ (quedan $\{3, 4\}$):
  $$\begin{vmatrix} 0 & -G_2 \\ -G_1 & G_1+G_2+G_3 \end{vmatrix} = 0 - ((-G_2)(-G_1)) = -G_1 G_2$$

Denominador ($[Y_{\text{ind}}]_{32}^{32}$): Se suprimen las filas $\{3, 2\}$ y las columnas $\{3, 2\}$ (quedan filas y columnas $\{1, 4\}$):
  $$\begin{vmatrix} G_2 & -G_2 \\ -G_2 & G_1+G_2+G_3 \end{vmatrix} = G_2(G_1+G_2+G_3) - (-G_2)^2 = G_2(G_1+G_3)$$

$$\frac{V_{12}}{V_{32}} = \frac{G_1 G_2}{G_2(G_1+G_3)} = \frac{G_1}{G_1+G_3}= \frac{s^2 LC}{s^2 LC + 1}$$

```python
# 1. Intentar importar o instalar pytc2 automáticamente
try:
    from pytc2.cuadripolos import calc_MAI_vtransf_ij_mn
    from pytc2.general import print_latex
except ModuleNotFoundError:
    import sys
    import subprocess
    print("Instalando pytc2...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "pytc2"])
    from pytc2.cuadripolos import calc_MAI_vtransf_ij_mn
    from pytc2.general import print_latex

import sympy as sp

# 2. Definición de símbolos de conductancia
G1, G2, G3 = sp.symbols('G1 G2 G3', real=True, positive=True)

# 3. Armado de la Matriz de Admitancia Indefinida (MAI)
# Nodos:         0 (V1)    1 (V2)    2 (V3)             3 (Central)
Ymai = sp.Matrix([
    [            G2,        0,        0,                   -G2],
    [             0,       G3,        0,                   -G3],
    [             0,        0,       G1,                   -G1],
    [           -G2,      -G3,      -G1,          G1 + G2 + G3]
])

print("Matriz de Admitancia Indefinida:")
print_latex(r'[Y_{\text{ind}}] = ' + sp.latex(Ymai))

# 4. Obtención de las transferencias (Punto 2.a)
# a) T_13^23 = V_23 / V_13 -> Salida: (nodo 1 - nodo 2), Entrada: (nodo 0 - nodo 2)
T_13_23 = calc_MAI_vtransf_ij_mn(Ymai, 1, 2, 0, 2, verbose=False)
T_13_23 = sp.simplify(T_13_23)

print("Transferencia V23 / V13:")
print_latex(r'T_{13}^{23} = \frac{V_{23}}{V_{13}} = ' + sp.latex(T_13_23))

# b) T_31^21 = V_21 / V_31 -> Salida: (nodo 1 - nodo 0), Entrada: (nodo 2 - nodo 0)
T_31_21 = calc_MAI_vtransf_ij_mn(Ymai, 1, 0, 2, 0, verbose=False)
T_31_21 = sp.simplify(T_31_21)

print("Transferencia V21 / V31:")
print_latex(r'T_{31}^{21} = \frac{V_{21}}{V_{31}} = ' + sp.latex(T_31_21))

# c) T_32^12 = V_12 / V_32 -> Salida: (nodo 0 - nodo 1), Entrada: (nodo 2 - nodo 1)
T_32_12 = calc_MAI_vtransf_ij_mn(Ymai, 0, 1, 2, 1, verbose=False)
T_32_12 = sp.simplify(T_32_12)

print("Transferencia V12 / V32:")
print_latex(r'T_{32}^{12} = \frac{V_{12}}{V_{32}} = ' + sp.latex(T_32_12))

# 5. Comprobación de complementariedad (Punto 2.b)
print("Comprobación |T_a + T_b| = 1:")
comp = sp.simplify(T_13_23 + T_31_21)
print_latex(r'|T_{13}^{23} + T_{31}^{21}| = ' + sp.latex(comp))
```
