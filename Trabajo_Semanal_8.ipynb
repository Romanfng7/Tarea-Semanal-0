### Ejercicio 1

$$F(s) = \frac{s^4 + 4s^2 + 3}{s(s^2 + 2)} = \frac{s^4 + 4s^2 + 3}{s^3 + 2s}$$

## a. Forma Foster 1

Partiendo de la expresión inicial tomamos por consigna $F(s)=Z(s)$. Para este tipo de desarrollo el numerador no puede ser más grande que el denominador, así que sacamos un factor común por división:

$$
\begin{array}{r|l}
s^4 + 4s^2 + 3 & \underline{\; s^3 + 2s \;} \\
\underline{-\,(s^4 + 2s^2)\,} & s \implies L_\infty = 1\text{ H} \\
2s^2 + 3 &
\end{array}
$$

Esto nos da de resultado:
$$Z(s) = F(s) = \frac{s^4 + 4s^2 + 3}{s(s^2 + 2)} = \frac{s^4 + 4s^2 + 3}{s^3 + 2s} = s + \frac{2s^2 + 3}{s(s^2 + 2)}$$

Ahora ya podemos hacer el despeje clásico de valores por límites:

Planteamos la descomposición para el resto:
$$Z(s) = s + \frac{A}{s} + \frac{B \cdot s}{s^2 + 2}$$

$$A = \lim_{s \to 0} \left[ s \cdot \frac{2s^2 + 3}{s(s^2 + 2)} \right] = \lim_{s \to 0} \frac{2s^2 + 3}{s^2 + 2} = \frac{2(0) + 3}{0 + 2} = \frac{3}{2}$$

$$B = \lim_{s^2 \to -2} \left[ \frac{s^2 + 2}{s} \cdot \frac{2s^2 + 3}{s(s^2 + 2)} \right] = \lim_{s^2 \to -2} \frac{2s^2 + 3}{s^2} = \frac{2(-2) + 3}{-2} = \frac{-1}{-2} = \frac{1}{2}$$

#### Expresión final Foster 1
$$Z(s) = s + \frac{3/2}{s} + \frac{\frac{1}{2}s}{s^2 + 2}$$

Por lo que podemos decir que es idéntica a la expresión matemática canónica:
$$F(s) = \frac{K_0}{s} + K_\infty s + \sum \frac{2 K_i s}{s^2 + \omega_i^2}$$

Acomodando la expresión nos quedaría:
$$Z(s) = \frac{3}{2}\frac{1}{s} + s + \frac{2\left(\frac{1}{4}\right)s}{s^2 + (\sqrt{2})^2}$$

---

### Deducción y cálculo de los componentes del circuito

Para ver claramente de dónde nacen los componentes, analizamos cada término de la suma en serie:

1. **Polo en el infinito (Inductor en serie):**
   $$Z_\infty(s) = K_\infty s = s \cdot L_\infty \implies L_\infty = 1\text{ H}$$

2. **Polo en el origen (Capacitor en serie):**
   $$Z_0(s) = \frac{K_0}{s} = \frac{1}{s \cdot \left(\cfrac{1}{K_0}\right)} = \frac{1}{s C_0}$$
   Como $K_0 = A = \frac{3}{2}$:
   $$C_0 = \frac{1}{K_0} = \frac{1}{3/2} = \frac{2}{3}\text{ F}$$

3. **Tanque antirresonante paralelo en serie (Pares de polos conjugados):**
   Invertimos algebraicamente el término genérico de impedancia $Z_i(s)$ para llevarlo a la forma de una admitancia en paralelo $Y_{\text{tanque}_i}(s) = s C_i + \frac{1}{s L_i}$[cite: 10]:
   $$Z_i(s) = \frac{2K_i s}{s^2 + \omega_i^2} = \frac{1}{\cfrac{s^2 + \omega_i^2}{2K_i s}} = \frac{1}{s \cdot \left(\cfrac{1}{2K_i}\right) + \cfrac{1}{s \cdot \left(\cfrac{2K_i}{\omega_i^2}\right)}}$$

   Comparando directamente el denominador[cite: 10]:
   $$C_i = \frac{1}{2K_i} \qquad \text{y} \qquad L_i = \frac{2K_i}{\omega_i^2}$$

   Aplicando para nuestro tanque con $2K_1 = B = \frac{1}{2}$ y $\omega_1^2 = 2$:
   $$Z_1(s) = \frac{\frac{1}{2}s}{s^2 + 2} = \frac{1}{\cfrac{s^2 + 2}{\frac{1}{2}s}} = \frac{1}{s \cdot \left(\cfrac{1}{1/2}\right) + \cfrac{1}{s \cdot \left(\cfrac{1/2}{2}\right)}} = \frac{1}{s(2) + \cfrac{1}{s(1/4)}}$$

   Por lo tanto:
   $$C_1 = \frac{1}{1/2} = 2\text{ F} \quad , \quad L_1 = \frac{1/2}{2} = \frac{1}{4}\text{ H}$$

Podemos concluir que los componentes que conforman el circuito Foster 1 en serie son:
* $L_\infty = 1\text{ H}$
* $C_0 = \frac{2}{3}\text{ F}$
* Tanque antirresonante paralelo: $C_1 = 2\text{ F}$ y $L_1 = \frac{1}{4}\text{ H}$

## b. Forma Foster 2

Para este punto tomamos por consigna la síntesis de admitancia en su versión derivación (paralelo), por lo que invertimos la función inmitancia inicial:

$$Y(s) = \frac{1}{Z(s)} = \frac{s(s^2 + 2)}{s^4 + 4s^2 + 3} = \frac{s^3 + 2s}{s^4 + 4s^2 + 3}$$

Factorizando el denominador (polinomio bicuadrático):
$$s^4 + 4s^2 + 3 = (s^2 + 1)(s^2 + 3)$$

Nos queda la función expresada como:
$$Y(s) = \frac{s^3 + 2s}{(s^2 + 1)(s^2 + 3)}$$

En este caso, como el grado del numerador (3) es menor que el del denominador (4), no hace falta realizar división polinómica ya que no existe término constante ni polo en el infinito ($K_\infty = 0$ y $K_0 = 0$).

Ahora ya podemos hacer el despeje clásico de valores por límites:

Planteamos la descomposición en fracciones simples para polos complejos conjugados sobre el eje imaginario:
$$Y(s) = \frac{A \cdot s}{s^2 + 1} + \frac{B \cdot s}{s^2 + 3}$$

$$A = \lim_{s^2 \to -1} \left[ \frac{s^2 + 1}{s} \cdot \frac{s(s^2 + 2)}{(s^2 + 1)(s^2 + 3)} \right] = \lim_{s^2 \to -1} \frac{s^2 + 2}{s^2 + 3} = \frac{-1 + 2}{-1 + 3} = \frac{1}{2}$$

$$B = \lim_{s^2 \to -3} \left[ \frac{s^2 + 3}{s} \cdot \frac{s(s^2 + 2)}{(s^2 + 1)(s^2 + 3)} \right] = \lim_{s^2 \to -3} \frac{s^2 + 2}{s^2 + 1} = \frac{-3 + 2}{-3 + 1} = \frac{-1}{-2} = \frac{1}{2}$$

#### Expresión final Foster 2
$$Y(s) = \frac{\frac{1}{2}s}{s^2 + 1} + \frac{\frac{1}{2}s}{s^2 + 3}$$

Por lo que podemos decir que es idéntica a la expresión matemática canónica:
$$F(s) = \frac{K_0}{s} + K_\infty s + \sum \frac{2 K_i s}{s^2 + \omega_i^2}$$

Acomodando la expresión identificando cada residuo $2K_i$ nos quedaría:
$$Y(s) = \frac{2\left(\frac{1}{4}\right)s}{s^2 + 1^2} + \frac{2\left(\frac{1}{4}\right)s}{s^2 + (\sqrt{3})^2}$$

---

### Deducción y cálculo de los componentes de cada rama

Para ver claramente de dónde nacen los componentes, invertimos algebraicamente el término genérico de cada admitancia parcial $Y_i(s)$ para llevarlo a la forma de una impedancia de rama serie:

$$Y_i(s) = \frac{2K_i s}{s^2 + \omega_i^2} = \frac{1}{\cfrac{s^2 + \omega_i^2}{2K_i s}} = \frac{1}{s \cdot \left(\cfrac{1}{2K_i}\right) + \cfrac{1}{s \cdot \left(\cfrac{2K_i}{\omega_i^2}\right)}}$$

Comparando el denominador con la impedancia de una rama serie $Z_{\text{rama}_i}(s) = s L_i + \frac{1}{s C_i}$, se identifican directamente los componentes:
$$L_i = \frac{1}{2K_i} \qquad \text{y} \qquad C_i = \frac{2K_i}{\omega_i^2}$$

Aplicando esta deducción a cada término:

* **Rama 1** ($2K_1 = \frac{1}{2}$, $\omega_1^2 = 1$):
  $$Y_1(s) = \frac{\frac{1}{2}s}{s^2 + 1} = \frac{1}{\cfrac{s^2 + 1}{\frac{1}{2}s}} = \frac{1}{s \cdot \left(\cfrac{1}{1/2}\right) + \cfrac{1}{s \cdot \left(\cfrac{1/2}{1}\right)}} = \frac{1}{s(2) + \cfrac{1}{s(1/2)}}$$
  $$\implies L_1 = 2\text{ H} \quad , \quad C_1 = \frac{1}{2}\text{ F}$$

* **Rama 2** ($2K_2 = \frac{1}{2}$, $\omega_2^2 = 3$):
  $$Y_2(s) = \frac{\frac{1}{2}s}{s^2 + 3} = \frac{1}{\cfrac{s^2 + 3}{\frac{1}{2}s}} = \frac{1}{s \cdot \left(\cfrac{1}{1/2}\right) + \cfrac{1}{s \cdot \left(\cfrac{1/2}{3}\right)}} = \frac{1}{s(2) + \cfrac{1}{s(1/6)}}$$
  $$\implies L_2 = 2\text{ H} \quad , \quad C_2 = \frac{1}{6}\text{ F}$$

## c.  Cauer con remoción en infinito (Cauer 1)

Para remover sucesivamente los polos en el infinito ($s \to \infty$), ordenamos dividendo y divisor en orden **decreciente** de potencias de $s$. Al comenzar desde una función de impedancia $Z(s)$, los cocientes alternan entre impedancias serie y admitancias en paralelo:

$$Z(s) = \frac{s^4 + 4s^2 + 3}{s^3 + 2s}$$

**Paso 1: Extracción de Impedancia (Inductor en serie)**
$$
\begin{array}{r|l}
s^4 + 4s^2 + 3 & \underline{\; s^3 + 2s \;} \\
\underline{-\,(s^4 + 2s^2)\;} & Z_1(s) = 1s \\
2s^2 + 3 &
\end{array}
$$
$$Z_1(s) = s \cdot L_1 = 1s \implies L_1 = 1\text{ H}$$

**Paso 2: Extracción de Admitancia (Capacitor en derivación)**
$$
\begin{array}{r|l}
s^3 + 2s & \underline{\; 2s^2 + 3 \;} \\
\underline{-\,(s^3 + \frac{3}{2}s)\;} & Y_2(s) = \frac{1}{2}s \\
\frac{1}{2}s &
\end{array}
$$
$$Y_2(s) = s \cdot C_2 = \frac{1}{2}s \implies C_2 = \frac{1}{2}\text{ F}$$

**Paso 3: Extracción de Impedancia (Inductor en serie)**
$$
\begin{array}{r|l}
2s^2 + 3 & \underline{\; \frac{1}{2}s \;} \\
\underline{-\,2s^2\;} & Z_3(s) = 4s \\
3 &
\end{array}
$$
$$Z_3(s) = s \cdot L_3 = 4s \implies L_3 = 4\text{ H}$$

**Paso 4: Extracción de Admitancia (Capacitor en derivación)**
$$
\begin{array}{r|l}
\frac{1}{2}s & \underline{\; 3 \;} \\
\underline{-\,\frac{1}{2}s\;} & Y_4(s) = \frac{1}{6}s \\
0 &
\end{array}
$$
$$Y_4(s) = s \cdot C_4 = \frac{1}{6}s \implies C_4 = \frac{1}{6}\text{ F}$$

#### Expresión en fracción continua (Cauer 1)
$$Z(s) = 1s + \cfrac{1}{\frac{1}{2}s + \cfrac{1}{4s + \cfrac{1}{\frac{1}{6}s}}}$$

Dada la función de admitancia de entrada:
$$Y(s) = \frac{s^5 + 18s^3 + 48s}{6s^4 + 42s^2 + 48}$$

## 1. Extracción del capacitor en serie $C_1$

Invertimos la función original para obtener la impedancia $Z(s)$:
$$Z(s) = \frac{1}{Y(s)} = \frac{6s^4 + 42s^2 + 48}{s^5 + 18s^3 + 48s}$$

Ordenamos numerador y denominador en potencias crecientes (remoción en DC / Cauer 2) y realizamos la primera división:

$$
\begin{array}{r|l}
48 + 42s^2 + 6s^4 & \underline{\; 48s + 18s^3 + s^5 \;} \\
\underline{-\,(48 + 18s^2 + s^4)\;} & 1\cfrac{1}{s} = Z_1(s) \\
24s^2 + 5s^4 &
\end{array}
$$

Identificamos el componente:
$$Z_1(s) = \frac{1}{s \cdot C_1} = 1\frac{1}{s} \implies \mathbf{C_1 = 1\text{ F}}$$

La impedancia remanente hacia la derecha es:
$$Z_A(s) = \frac{24s^2 + 5s^4}{48s + 18s^3 + s^5}$$

---

## 2. Extracción del inductor en paralelo $L_1$

Como el siguiente elemento es un inductor en derivación $L_1$, invertimos $Z_A(s)$ a admitancia:
$$Y_A(s) = \frac{1}{Z_A(s)} = \frac{48s + 18s^3 + s^5}{24s^2 + 5s^4}$$

Realizamos la división en potencias crecientes para extraer el polo en el origen:

$$
\begin{array}{r|l}
48s + 18s^3 + s^5 & \underline{\; 24s^2 + 5s^4 \;} \\
\underline{-\,(48s + 10s^3)\;} & 2\cfrac{1}{s} = Y_1(s) \\
8s^3 + s^5 &
\end{array}
$$

Identificamos el componente:
$$Y_1(s) = \frac{1}{s \cdot L_1} = 2\frac{1}{s} \implies \mathbf{L_1 = \frac{1}{2}\text{ H} = 0.5\text{ H}}$$

La admitancia remanente queda conformada por el resto sobre el divisor:
$$Y_{\text{rem}}(s) = \frac{8s^3 + s^5}{24s^2 + 5s^4} = \frac{s^3(8 + s^2)}{s^2(24 + 5s^2)} = \frac{s(s^2 + 8)}{5s^2 + 24}$$

---

## 3. Síntesis del tanque $L_2 \parallel C_2$ y capacitor $C_3$ (Foster)

Invertimos la admitancia remanente a impedancia para analizar la rama serie restante:
$$Z_{\text{rem}}(s) = \frac{1}{Y_{\text{rem}}(s)} = \frac{5s^2 + 24}{s(s^2 + 8)}$$

Descomponemos en fracciones simples correspondientes a un polo en el origen ($C_3$) y un par de polos complejos conjugados ($L_2 \parallel C_2$):
$$Z_{\text{rem}}(s) = \frac{A}{s} + \frac{B \cdot s}{s^2 + 8}$$


$A = \lim_{s \to 0} \left[ s \cdot \frac{5s^2 + 24}{s(s^2 + 8)} \right] = \frac{24}{8} = 3$

$B = \lim_{s^2 \to -8} \left[ \frac{s^2 + 8}{s} \cdot \frac{5s^2 + 24}{s(s^2 + 8)} \right] = \lim_{s^2 \to -8} \frac{5s^2 + 24}{s^2} = \frac{5(-8) + 24}{-8} = \frac{-16}{-8} = 2$

Expresión de la impedancia remanente:
$$Z_{\text{rem}}(s) = \frac{3}{s} + \frac{2s}{s^2 + 8}$$

### Deducción de los componentes:

* **Capacitor $C_3$:**
$$Z_{C3}(s) = \frac{3}{s} = \frac{1}{s \cdot C_3} \implies \mathbf{C_3 = \frac{1}{3}\text{ F}}$$

* **Tanque antirresonante paralelo ($L_2 \parallel C_2$):**
Invertimos algebraicamente el término de impedancia para comparar con la admitancia de un tanque paralelo $Y_{\text{tanque}}(s) = s C_2 + \frac{1}{s L_2}$[cite: 10]:
$$Z_{\text{tanque}}(s) = \frac{2s}{s^2 + 8} = \frac{1}{\cfrac{s^2 + 8}{2s}} = \frac{1}{s \cdot \left(\cfrac{1}{2}\right) + \cfrac{1}{s \cdot \left(\cfrac{1}{4}\right)}}$$

Comparando término a término[cite: 10]:
$$\mathbf{C_2 = \frac{1}{2}\text{ F} = 0.5\text{ F}}$$
$$\mathbf{L_2 = \frac{1}{4}\text{ H} = 0.25\text{ H}}$$

---

## 4. Resumen final de componentes

* $C_1 = 1\text{ F}$[cite: 13]
* $L_1 = 0.5\text{ H}$[cite: 13]
* $C_2 = 0.5\text{ F}$
* $L_2 = 0.25\text{ H}$
* $C_3 = \frac{1}{3}\text{ F}$

$$Z(s) = \frac{s^2 + 6s + 8}{s^2 + 6s + 5}$$

$$
\begin{array}{r|l}
s^2 + 6s + 8 & \underline{\; s^2 + 6s + 5 \;} \\
\underline{s^2 + 6s + 5\;} & 1 \longrightarrow \text{Resistencia} = 1\,\Omega \\
3 &
\end{array}
$$

$$\frac{3}{s^2 + 6s + 5} = \frac{3}{(s+1)(s+5)}$$

$$\lim_{s \to -1} (s+1) \frac{3}{(s+1)(s+5)} = \frac{3}{4}$$

$$\lim_{s \to -5} (s+5) \frac{3}{(s+1)(s+5)} = \frac{3}{-4}$$

No puede ser negativo, tomamos otro enfoque analizando los bloques paralelo $RC$ y $RL$:

$$Z_2(s) = R_2 \parallel (s L_2) = \frac{R_2 \cdot s L_2}{s L_2 + R_2} = \frac{s R_2}{s + \frac{R_2}{L_2}}$$

$$Z_3(s) = R_3 \parallel \left(\frac{1}{s C_1}\right) = \frac{R_3 \cdot \frac{1}{s C_1}}{R_3 + \frac{1}{s C_1}} = \frac{\frac{1}{C_1}}{s + \frac{1}{R_3 C_1}}$$

Asignamos los polos del denominador:
$$\frac{1}{R_3 C_1} = 1$$
$$\frac{R_2}{L_2} = 5$$

Con el residuo del bloque $RC$:
$$Z_3(s) = \frac{3/4}{s + 1} \implies \frac{1}{C_1} = \frac{3}{4} \implies C_1 = \frac{4}{3}\text{ F}$$

$$R_3 = \frac{1}{C_1 \cdot 1} = \frac{3}{4}\,\Omega$$

Para el bloque $RL$, la impedancia remanente se obtiene restando $Z_3(s)$:
$Z_2(s) =Z_A(s) - Z_1(s) = \frac{3}{(s+1)(s+5)} - \frac{3/4}{s+1} = \frac{3 - \frac{3}{4}(s+5)}{(s+1)(s+5)} = \frac{-\frac{3}{4}s - \frac{3}{4}}{(s+1)(s+5)} = \frac{-\frac{3}{4}}{s+5}$

Reexpresando con la forma de $Z_2(s) = \frac{s R_2}{s + 5}$:
$$R_1' = R_1 - R_2$$
$$R_1 = 0.85\,\Omega$$
$$R_2 = 0.15\,\Omega$$
$$L_2 = \frac{R_2}{5} = \frac{0.15}{5} = 0.03\text{ H}$$

Resumen de componentes:
* $R_1 = 0.85\,\Omega$
* $R_2 = 0.15\,\Omega$
* $L_2 = 0.03\text{ H}$
* $R_3 = 0.75\,\Omega$
* $C_1 = 4/3\text{ F}$

```python
import sympy as sp
import inspect
from pytc2.sintesis_dipolo import foster, cauer_LC
from pytc2.dibujar import dibujar_foster_serie, dibujar_foster_derivacion, dibujar_cauer_LC
from pytc2.general import s, print_latex, print_subtitle
import pytc2.dibujar as _dib

sp.init_printing()

# --- Helper local: reemplaza a pytc2.general.a_equal_b_latex_s ---------
def a_eq_b(nombre, expr):
    b_str = sp.latex(expr) if isinstance(expr, (sp.Expr, sp.MatrixBase, list)) else expr
    return '$' + nombre + '=' + b_str + '$'

# --- Wrapper defensivo: soporta foster() con 3 o 5 valores de retorno,
#     y normaliza None -> 0 (versiones viejas de pytc2 usan None para
#     "sin término", en vez de sp.Integer(0)) ------------------------------
def foster_seguro(expr):
    res = foster(expr)
    if len(res) == 5:
        k0, koo, ki_wi, kk, F_foster = res
    elif len(res) == 3:
        k0, koo, ki_wi = res
        kk = None
        F_foster = None
    else:
        raise RuntimeError(f'foster() devolvió {len(res)} valores inesperados.')

    k0 = sp.Integer(0) if k0 is None else k0
    koo = sp.Integer(0) if koo is None else koo
    kk = sp.Integer(0) if kk is None else kk
    ki_wi = [] if ki_wi is None else ki_wi

    if F_foster is None:
        F_foster = k0/s + koo*s
        for ki_, wi_ in ki_wi:
            F_foster += 2*ki_*s/(s**2 + wi_**2)

    return [k0, koo, ki_wi, kk, F_foster]

# --- Parche defensivo: bug conocido en dibujar_foster_derivacion -------
# (compara "kk is None" en vez de "kk.is_zero" y dibuja resistencias
#  en vez de tanques LC puros; se corrige solo si el patrón está presente)
_src = inspect.getsource(_dib.dibujar_foster_derivacion)
_buggy = "if kk is None:\n            bDisipativo = False"
_fix = "if kk.is_zero:\n            bDisipativo = False"
if _buggy in _src:
    _src_corregido = _src.replace(_buggy, _fix)
    _ns = dict(_dib.__dict__)
    exec(compile(_src_corregido, '<dibujar_foster_derivacion corregida>', 'exec'), _ns)
    dibujar_foster_derivacion_fixed = _ns['dibujar_foster_derivacion']
    print('Parche aplicado a dibujar_foster_derivacion.')
else:
    dibujar_foster_derivacion_fixed = dibujar_foster_derivacion
    print('No se detectó el bug conocido; se usa dibujar_foster_derivacion original.')

# =========================================================================
# 1. Función de excitación
# =========================================================================
F = (s**4 + 4*s**2 + 3) / (s*(s**2 + 2))
F = sp.simplify(F)
F_factorizada = sp.factor(sp.numer(F)) / sp.factor(sp.denom(F))

print_latex(a_eq_b('F(s)', F))
print_latex(a_eq_b('F(s)', F_factorizada))
display(F)

# =========================================================================
# 2. Verificación: ¿es una función de reactancia LC válida?
# =========================================================================
es_impar = sp.simplify(F.subs(s, -s) + F) == 0
print('¿F(s) es impar? ->', es_impar)

ceros = sp.solve(sp.numer(F_factorizada), s)
polos = sp.solve(sp.denom(F_factorizada), s)
print('Ceros de F(s):', ceros)
print('Polos de F(s):', polos)

w_ceros = sorted({sp.Abs(sp.im(z)) for z in ceros})
w_polos = sorted({sp.Abs(sp.im(p)) if p != 0 else sp.Integer(0) for p in polos})
print('Frecuencias de los ceros (>=0):', w_ceros)
print('Frecuencias de los polos (>=0):', w_polos)

todas_reales_o_imaginarias = all(sp.im(r) == 0 or sp.re(r) == 0 for r in (ceros + polos))
print('¿Todos los polos/ceros están sobre el eje jw?', todas_reales_o_imaginarias)

secuencia = sorted([(w, 'polo') for w in w_polos] + [(w, 'cero') for w in w_ceros])
print('Secuencia polo/cero creciente en w:', secuencia)

# =========================================================================
# 3. Foster I - forma serie ( Z(s) = F(s) )
# =========================================================================
print_subtitle('Foster I (serie)')

k0, koo, ki_wi, kk, F_foster = foster_seguro(F)

print('k0  =', k0, '  -> capacitor serie C0 = 1/k0 =', sp.nsimplify(1/k0) if k0 != 0 else '∞', 'F')
print('koo =', koo, '  -> inductor serie Loo = koo =', koo, 'H')
print('tanques LC serie (1/ki, ci):', ki_wi)
for tanque in ki_wi:
    print('   -> L =', sp.nsimplify(1/tanque[0]), 'H  ,  C =', tanque[1], 'F')

print('F_foster =', F_foster)
print('Verificación F_foster - F =', sp.simplify(F_foster - F), ' (debe ser 0)')

dibujar_foster_serie(k0, koo, ki_wi, z_exc=F)

# =========================================================================
# 4. Foster II - forma derivación ( Y(s) = 1/F(s) )
# =========================================================================
print_subtitle('Foster II (derivación)')

Y = sp.simplify(1/F)
print_latex(a_eq_b('Y(s) = 1/F(s)', Y))

k0y, kooy, ki_wiy, kky, Y_foster = foster_seguro(Y)

print('k0y  =', k0y)
print('kooy =', kooy)
print('ramas LC serie en derivación (ki, wi):', ki_wiy)
for rama in ki_wiy:
    print('   -> L =', sp.nsimplify(rama[1]), 'H  ,  C =', sp.nsimplify(1/rama[0]), 'F')

print('Y_foster =', Y_foster)
print('Verificación Y_foster - Y =', sp.simplify(Y_foster - Y), ' (debe ser 0)')

dibujar_foster_derivacion_fixed(k0y, kooy, ki_wiy, y_exc=Y)

# =========================================================================
# 5. Cauer I - remoción sucesiva en s -> infinito
# =========================================================================
print_subtitle('Cauer I (remoción en infinito)')

koo_c, F_cauer_oo, rem_oo = cauer_LC(F, remover_en_inf=True)

print('Coeficientes de la fracción continua (orden: serie, derivación, serie, ...):')
for i, ki in enumerate(koo_c):
    valor = sp.nsimplify(ki/s)
    elemento = 'L (serie)' if i % 2 == 0 else 'C (derivación)'
    print(f'   elemento {i+1}: {elemento} = {valor}  [{ki}]')

print('F_cauer_oo =', F_cauer_oo)
print('Resto de la expansión:', rem_oo, ' (debe ser 0 para expansión exacta)')
print('Verificación F_cauer_oo - F =', sp.simplify(F_cauer_oo - F), ' (debe ser 0)')

dibujar_cauer_LC(koo_c, z_exc=F_cauer_oo)

# =========================================================================
# 6. Cauer II - remoción sucesiva en s -> 0 (continua)
# =========================================================================
print_subtitle('Cauer II (remoción en continua)')

k0_c, F_cauer_0, rem_0 = cauer_LC(F, remover_en_inf=False)

print('Coeficientes de la fracción continua (orden: serie, derivación, serie, ...):')
for i, ki in enumerate(k0_c):
    valor = sp.nsimplify(1/(s*ki))
    elemento = 'C (serie)' if i % 2 == 0 else 'L (derivación)'
    print(f'   elemento {i+1}: {elemento} = {valor}  [{ki}]')

print('F_cauer_0 =', F_cauer_0)
print('Resto de la expansión:', rem_0, ' (debe ser 0 para expansión exacta)')
print('Verificación F_cauer_0 - F =', sp.simplify(F_cauer_0 - F), ' (debe ser 0)')

dibujar_cauer_LC(k0_c, z_exc=F_cauer_0)

# =========================================================================
# 7. Resumen
# =========================================================================
print_subtitle('Resumen')
print('Foster I:  C0=2/3F -- Loo=1H -- tanque(L=1/4H || C=2F), todo en serie')
print('Foster II: rama(L=2H,C=1/2F) || rama(L=2H,C=1/6F)')
print('Cauer I:   L1=1H -- C2=1/2F -- L3=4H -- C4=1/6F  (escalera)')
print('Cauer II:  C1=2/3F -- L2=5/4H -- C3=2/25F -- L4=5H  (escalera)')
```
