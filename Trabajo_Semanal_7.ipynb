**Segundo bloque (Red recursiva de 2do orden / Biquad):**

1.   $W(z) = V(z) \cdot a_0$
2.   $V(z) = \frac{W(z)}{a_0}$

$v(n) = x_3(n) + w(n-1)a_1 + w(n-2)a_2$.  
1.   $V(z) = X_3(z) + W(z)z^{-1}a_1 + W(z)z^{-2}a_2$

3.   $\frac{W(z)}{a_0} - W(z)z^{-1}a_1 - W(z)z^{-2}a_2 = X_3(z)$
4.   $W(z) \left( \frac{1}{a_0} - z^{-1}a_1 - z^{-2}a_2 \right) = X_3(z)$

$y(n) = w(n)b_0 + w(n-1)b_1 + w(n-2)b_2$.  
1.   $Y(z) = W(z)b_0 + W(z)z^{-1}b_1 + W(z)z^{-2}b_2$

$$\frac{Y(z)}{X_3(z)} = \frac{b_0 + z^{-1}b_1 + z^{-2}b_2}{\frac{1}{a_0} - z^{-1}a_1 - z^{-2}a_2}$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fftpack import diff

# ==========================================
# Generación de la señal de prueba
# ==========================================
# Elegimos un omega que genere un periodo entero de muestras
omega = 0.1 * np.pi
periodo_muestras = 2 * np.pi / omega  # 20 muestras por ciclo

# Generamos exactamente 5 ciclos (100 muestras)
n = np.arange(100)
x = np.sin(omega * n)

# ==========================================
# 1. Derivada analítica (Referencia ideal)
# ==========================================
# d/dn [sin(w*n)] = w * cos(w*n)
y_analitica = omega * np.cos(omega * n)

# ==========================================
# 2. Filtro diferenciador discreto (Punto 3b)
# ==========================================
y_filtro = np.zeros_like(x)
for i in range(2, len(n)):
    y_filtro[i] = 0.5 * x[i] - 0.5 * x[i-2]

# Compensamos el retardo de grupo constante de 1 muestra para alinear las fases
y_filtro_compensada = np.roll(y_filtro, -1)
y_filtro_compensada[-1] = 0 # Limpiamos la última muestra desplazada

# ==========================================
# 3. Derivada por FFT (Punto Bonus)
# ==========================================
# Se pasa el periodo real de la secuencia para escalar correctamente los coeficientes de Fourier
y_fft = diff(x, order=1, period=periodo_muestras)

# ==========================================
# Visualización de resultados
# ==========================================
plt.figure(figsize=(10, 5))
plt.plot(n, y_analitica, label=r'Analítica: $\Omega \cos(\Omega n)$',
         color='black', linestyle='dashed', linewidth=3, alpha=0.6)

plt.plot(n, y_fft, label='scipy.fftpack.diff (Bonus)',
         color='red', marker='x', markersize=6, linestyle='none')

plt.plot(n, y_filtro_compensada, label='Filtro en diferencias (Compensado)',
         color='blue', marker='o', markersize=4, alpha=0.7)

plt.xlim(0, 40) # Hacemos zoom en los primeros dos ciclos
plt.title('Comparación de métodos de derivación en tiempo discreto')
plt.xlabel('Muestras (n)')
plt.ylabel('Amplitud')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as signal

# ==========================================
# Función auxiliar para graficar Bode y Plano Z
# ==========================================
def plot_filter_analysis(b, a, title):
    w, h = signal.freqz(b, a, worN=1024)
    w_pi = w / np.pi
    mag = 20 * np.log10(np.abs(h) + 1e-12)
    phase = np.angle(h)
    z, p, k = signal.tf2zpk(b, a)

    fig = plt.figure(figsize=(15, 4))
    fig.suptitle(title, fontsize=14, fontweight='bold')

    # Módulo
    ax1 = fig.add_subplot(131)
    ax1.plot(w_pi, mag, 'b')
    ax1.set_title('Respuesta de Módulo')
    ax1.set_xlabel(r'Frecuencia (x $\pi$ rad/muestra)')
    ax1.set_ylabel('Magnitud (dB)')
    ax1.grid(True)

    # Fase
    ax2 = fig.add_subplot(132)
    ax2.plot(w_pi, phase, 'g')
    ax2.set_title('Respuesta de Fase')
    ax2.set_xlabel(r'Frecuencia (x $\pi$ rad/muestra)')
    ax2.set_ylabel('Fase (radianes)')
    ax2.grid(True)

    # Plano Z
    ax3 = fig.add_subplot(133)
    unit_circle = plt.Circle((0, 0), 1, color='black', fill=False, linestyle='--')
    ax3.add_patch(unit_circle)
    ax3.scatter(np.real(z), np.imag(z), s=60, marker='o', facecolors='none', edgecolors='b', label='Ceros')
    ax3.scatter(np.real(p), np.imag(p), s=60, marker='x', color='r', label='Polos')
    ax3.set_xlim([-1.5, 1.5])
    ax3.set_ylim([-1.5, 1.5])
    ax3.set_aspect('equal')
    ax3.axhline(0, color='black', lw=0.5)
    ax3.axvline(0, color='black', lw=0.5)
    ax3.set_title('Plano Z')
    ax3.grid(True, linestyle=':')
    ax3.legend(loc='upper right')

    plt.tight_layout()
    plt.show()

# ==========================================
# Inciso 2: Filtro CIC (Media Móvil N=4)
# ==========================================
print("--- Ejercicio 2: Filtro CIC (N=4) ---")
N_cic = 4
b_cic = np.zeros(N_cic + 1)
b_cic[0] = 1/N_cic
b_cic[N_cic] = -1/N_cic
a_cic = [1, -1]
plot_filter_analysis(b_cic, a_cic, 'Filtro CIC (N=4)')

# ==========================================
# Inciso 3: Filtro Diferenciador
# ==========================================
print("--- Ejercicio 3: Filtro Diferenciador ---")
b_diff = [0.5, 0.0, -0.5]
a_diff = [1.0]
plot_filter_analysis(b_diff, a_diff, 'Filtro Diferenciador')

# ==========================================
# Inciso 4: Filtro Elimina Continua (DC Blocker)
# ==========================================
print("--- Ejercicio 4: Filtro Elimina Continua (DC Blocker) ---")
beta = 0.7265
b_dc = [1.0, -1.0]
a_dc = [1.0, -beta]
plot_filter_analysis(b_dc, a_dc, r'DC Blocker ($\beta = 0.7265$)')

# ==========================================
# Inciso 5: Filtro Peine (Comb filter)
# ==========================================
print("--- Ejercicio 5: Filtro Peine (N=8) ---")
b_comb = np.zeros(9)
b_comb[0] = 1.0
b_comb[8] = 1.0  # + z^-8
a_comb = [1.0]
plot_filter_analysis(b_comb, a_comb, 'Filtro Peine (N=8)')
```
